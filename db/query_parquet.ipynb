# Query GitSkills Parquet files with DuckDB

Queries run directly on the Parquet files; nothing is copied or imported. Each dataset is exposed as a view so queries can use short names.

```python
from pathlib import Path
import duckdb

# Find the repo root (the folder that holds pyproject.toml), wherever the notebook is run from
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
DATA = ROOT / "src" / "data" / "gitskills_data" / "data"

con = duckdb.connect()  # in-memory; views only store the query, not the data
for name in ["repos", "artifacts", "artifact_siblings", "mining_runs"]:
    con.sql(f"CREATE VIEW {name} AS SELECT * FROM read_parquet('{DATA / name}/*.parquet')")

con.sql("SHOW TABLES")
```

## Sample query

The 10 most common SKILL file names across all artifacts.

```python
con.sql("""
    SELECT filename, COUNT(*) AS variant_count
    FROM artifacts
    GROUP BY filename
    ORDER BY variant_count DESC
    LIMIT 10
""").df()
```
